# Data Audit

This notebook audits the raw workbook, checks schema expectations, summarizes missingness, and shows validation flags. The canonical scientific table is the `Combined_Team13` worksheet.

```python
from pathlib import Path
from laserdb.io import load_workbook, load_laser_data
from laserdb.clean import build_processed_table
from laserdb.validate import missingness_summary, find_parse_failures, validate_database

workbook = Path('../combined_team13_format.xlsx')
sheets = load_workbook(workbook)
sheets.keys()
```

```python
raw = load_laser_data(workbook)
raw.head()
```

```python
raw.dtypes
```

```python
processed = build_processed_table(raw)
missingness_summary(processed)
```

```python
find_parse_failures(processed)
```

```python
report = validate_database(processed)
{k: report[k] for k in ['unrecognized_laser_types', 'unrecognized_demonstrated_status', 'missing_source_count', 'parse_failure_count']}
```

## Schema Notes

Raw FOM columns are preserved. Numeric parsing creates separate `*_num` columns and display-unit columns such as `peak_power_pw` and `wavelength_nm`. Validation flags incomplete or suspicious records for curation but does not modify raw scientific values.
