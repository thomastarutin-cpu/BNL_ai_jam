# Exploratory Plots

This notebook uses the processed table to make interactive plots. Log scales are often useful because laser peak power, wavelength, repetition rate, and intensity span many orders of magnitude. Missing FOM values are left missing and are not plotted as zero.

```python
import pandas as pd
from laserdb.plotting import (
    peak_power_vs_duration,
    peak_power_vs_rep_rate,
    wavelength_vs_peak_power,
    peak_power_histogram,
    counts_by_country,
    counts_by_laser_type,
    missingness_heatmap,
)

df = pd.read_csv('../data/processed/laser_data_processed.csv')
df.head()
```

Peak power versus pulse duration helps separate ultrashort petawatt systems from longer-pulse systems.

```python
peak_power_vs_duration(df)
```

Peak power versus repetition rate highlights systems that combine high pulse energy with frequent operation.

```python
peak_power_vs_rep_rate(df)
```

Wavelength versus peak power distinguishes optical laser entries from FEL and x-ray records.

```python
wavelength_vs_peak_power(df)
```

```python
peak_power_histogram(df)
```

```python
counts_by_country(df)
```

```python
counts_by_laser_type(df)
```

The missingness heatmap shows which numeric FOM values could be parsed for each record.

```python
missingness_heatmap(df)
```
